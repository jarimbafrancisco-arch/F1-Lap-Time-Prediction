# F1 Lap Time Prediction - Data Download and Inspection

This project aims to predict a Formula 1 driver's next lap time using information available at the end of the current lap. I will explore previous lap times, tyre conditions and weather to prepare the data for a machine learning regression model.

In this notebook, I use the FastF1 Python library to download and inspect race data, starting with Monza 2024. This helps me understand the available data before moving on to exploratory analysis.

## Imports

```python
import fastf1
```

## Event selection

```python
schedule=fastf1.get_event_schedule(2024, include_testing=False)
schedule[["RoundNumber", "EventName", "EventDate"]]
```

## Race Selection

```python
session=fastf1.get_session(2024, "Monza", "R")
session.event[["EventName", "EventDate"]]
```

## Session Load

```python
session.load(telemetry=False)
laps=session.laps
laps.head()
```

## View lap data in a separate tab

```python
import pandas as pd

laps_view = pd.DataFrame(laps).copy()
laps_view.head()
```

## Data that will be used

```python
seasons=range(2022,2026)
schedules=[]

for year in seasons:
    schedule=fastf1.get_event_schedule(year, include_testing=False)
    schedule["Season"]=year
    schedules.append(schedule)

events=pd.concat(schedules,ignore_index=True)
events.groupby("Season")["RoundNumber"].count()
```

## Saving Data

```python
from pathlib import Path

project_root=Path.cwd()
if project_root.name=="notebooks":
    project_root=project_root.parent

data_dir=project_root / "data"
data_dir.mkdir(exist_ok=True)
events.to_csv(data_dir / "events_2022_2025.csv", index=False)
```
